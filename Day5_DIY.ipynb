```python
#Q1. Problem Statement: Math Quiz
'''Write a Python program to create a math quiz that generates math questions
using random numbers and operators and asks the user to enter the correct
answer; evaluate the answer entered by the user and print a message whether
the answer is correct or incorrect.'''
```

```python
import random
name=input("What is your name:")
print(f"Alright {name} Welcome to your maths quiz")
a=random.randint(1,10)
b=random.randint(1,10)
c=["+","-","/","*"]
d=random.choice(c)
maths=eval(str(a)+d+str(b))
#The eval() allows us to execute arbitrary strings as Python code. 
#It accepts a source string and returns an object.
print("Your question is",a,d,b)
answer=float(input("what is your answer:"))
if answer==maths:
    print("Answer is correct",maths)
else:
    print("Incorrect. The correct answer is:",maths)
```

```python
#Q2. Problem Statement: Understanding Classes and Objects
'''Write a Python program to create a Bus object that will inherit all the variables
and methods of the parent Vehicle class and display it.'''
```

```python
class parent_vehicle():
    def __init__(self,name,speed):
        self.name=name
        self.speed=speed
    def info(self):
        return f"Vehicle name:{self.name} Vehicle speed: {self.speed}"
        
class Bus(parent_vehicle):
    def __init__(self,name,speed,milege):
        super().__init__(name,speed)
        self.milege=milege
    def info(self):
        return f"{super().info()} Vehicle Milege: {self.milege}"

bus=Bus("Toyota","157km/h","29km/l")
print(bus.info())
```
